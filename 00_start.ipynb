```python
import torch.nn as nn
import torch.nn.functional as F
# Create a model class that inherits nn.Module
class Model(nn.Module):
  # input layer with 4 feature flowers
  def __init__(self,in_features=4,h1=8,h2=9,out_feature=3 ,*args, **kwargs) -> None:
    super().__init__(*args, **kwargs)
    self.fc1 = nn.Linear(in_features,h1)
    self.fc2 = nn.Linear(h1,h2)
    self.out = nn.Linear(h2,out_feature)

  def forward(self, x):
    x = F.relu(self.fc1(x))
    x = F.relu(self.fc2(x))
    x = self.out(x)
    return x
torch.manual_seed(41)
model = Model()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
%matplotlib inline
url = 'https://gist.githubusercontent.com/netj/8836201/raw/6f9306ad21398ea43cba4f7d537619d0e07d5ae3/iris.csv'
my_df = pd.read_csv(url)
my_df.head()
```

```python
my_df['variety'] = my_df['variety'].replace('Setosa', 0.0)
my_df['variety'] = my_df['variety'].replace('Versicolor', 1.0)
my_df['variety'] = my_df['variety'].replace('Virginica', 2.0)
my_df.head()
```

```python
x = my_df.drop('variety',axis=1).values
y = my_df['variety'].values
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=41)

X_train = torch.FloatTensor(x_train)
X_test = torch.FloatTensor(x_test)
y_train = torch.LongTensor(y_train)
y_test = torch.LongTensor(y_test)
```

```python
criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(model.parameters(),lr=0.01)

epochs = 100
losses = []
for i in range (epochs):
  # go forward and get a predicction
  y_pred = model.forward(X_train)

  # measure the loss/Error, gonna be high at first
  loss = criterion(y_pred, y_train)

  # keep track of our losses
  losses.append(loss.detach().numpy())

  if i % 10 == 0:
    print(f'Epoch : {i} and loss : {loss}')

  # do some back propagation: take error rate of forward propagation
  optimizer.zero_grad() # learn this later
  loss.backward()
  optimizer.step() # Updates model parameters using the computed gradients and the optimization algorithm (Adam).
```

```python
# graph it out
plt.plot(range(epochs),losses)
plt.ylabel('Loss')
plt.xlabel('Epoch')
```

```python
# evaluate model on test data set (validate model on test set)
with  torch.no_grad(): # basicall turn of back propagation
  y_eval = model.forward(X_test)
  loss = criterion(y_eval,y_test)
print(f'The loss on the test set is : {loss}')

```

```python
correct = 0

with torch.no_grad():
  for i, data in enumerate(X_test):
    y_val = model.forward(data)
    print(f'{i+1} {str(y_val)} {y_test[i]} \t {y_val.argmax().item()}')
    if y_val.argmax().item() == y_test[i]:
      correct += 1
print(f'{correct} out of {len(y_test)} = {100*correct/len(y_test)}% correct')
```
